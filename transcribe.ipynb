```python
from faster_whisper import WhisperModel
import os
import subprocess
```

```python
# Configuration
input_audio_path = '/Users/vlad/Library/CloudStorage/GoogleDrive-vladimir.smirnov@rhodaris.com/Shared drives/rhodaris working drive/Courses PMP maintenance 2026/Navigating Tomorrow - New Insights on the AI-Driven Project Management Revolution.m4a'
output_audio_name = "Navigating Tomorrow - New Insights on the AI-Driven Project Management Revolution_output16khz.wav"
output_transcription_name = "Navigating Tomorrow - New Insights on the AI-Driven Project Management Revolution transcription.txt"
meeting_language = "en"

# Setup paths
folder_path = os.path.dirname(input_audio_path)
output_audio_path = os.path.join(folder_path, output_audio_name)
output_file_path = os.path.join(folder_path, output_transcription_name)
```

```python
# Load Whisper Model (optimized for Mac M4)
model = WhisperModel(
    "large-v2", 
    # "medium",
    device="cpu", 
    compute_type="int8",    
    num_workers=1,
    cpu_threads=8
)
print("compute type:", {model.model.compute_type})
```

```python
# Preprocess audio with ffmpeg (16kHz, mono)
subprocess.run([
    "ffmpeg", "-y", "-vn", "-i", input_audio_path, 
    "-ar", "16000", "-ac", "1", output_audio_path
], check=True)

# print(f"Audio preprocessed: {output_audio_path}")
```

```python
# Transcribe Audio
segments, info = model.transcribe(
    output_audio_path, 
    language=meeting_language
)

transcription = " ".join([segment.text for segment in segments])

print(f"Detected language: {info.language} (probability: {info.language_probability:.2f})")
print(f"Duration: {info.duration:.2f}s")
```

```python
# Save transcription
with open(output_file_path, "w", encoding="utf-8") as file:
    file.write(transcription)

print(f"Transcription saved to: {output_file_path}")
```
