```python
# read text file
file_path = "/Users/vlad/Dev/files/transcription 2.txt"  # Specify the output file path

```

```python
from wordcloud import WordCloud, STOPWORDS
import matplotlib.pyplot as plt

# Исходный текст
with open(file_path, "r") as file:
    text = file.read()
# print(text)
# Пользовательские русские стоп-слова
russian_stopwords = {
    "что", "это", "я", "в", "на", "и", "а", "ну", "там", "вот", "бы", "все", "не", "то",
    "как", "ну", "с", "по", "за", "до", "у", "от", "для", "о", "так", "но", "да", "же", "или",
    "к", "из", "еще", "где", "когда", "тут", "он", "она", "они", "мы", "ты", "вы", "быть", "если","чтоб", 
    "наверное","чтобы","мне","нибудь","стать","конечно","надо","было","чему","опять","таки","зачем", "все", "всегда", "всё", "всякий", "всяко", "вся", "всё", "всех", "всем", "всеми", "всеми", "всему",
    "соответственно", "прям", "довольно", "чего", "го","какой", "вообще","принципе", "предполагаю", "сессий","банально", "поэтому","исследование","хрена", "точка",
    "сказал", "знаю",
    "меня", "блин", "помню", "говорю", "какие", "того", "кто", "тоже", "чем"
    }

# Объединяем с дефолтными стоп-словами
custom_stopwords = set(STOPWORDS).union(russian_stopwords)

# Генерация облака слов
wordcloud = WordCloud(
    width=800,
    height=400,
    background_color='white',
    stopwords=custom_stopwords,
    font_path='/System/Library/Fonts/Supplemental/Arial Unicode.ttf'  # путь к шрифту, поддерживающему кириллицу
).generate(text)

# Отображение
plt.figure(figsize=(12, 6))
plt.imshow(wordcloud, interpolation='bilinear')
plt.axis('off')
plt.tight_layout()
plt.show()
```
