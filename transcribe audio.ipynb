# Import Required Libraries
Import the necessary libraries, including the Whisper library and any other dependencies.

```python
# Import Required Libraries
import whisper
import os
import imageio_ffmpeg as ffmpeg
import subprocess
```

# Load Whisper Model
Load the Whisper model using the appropriate function from the Whisper library.

```python
# Load Whisper Model
# model = whisper.load_model("base")
model = whisper.load_model("large-v2")
```

Preprocess the Audio: Use ffmpeg to preprocess the audio file (e.g., normalize volume, resample to 16 kHz):

mpeg -i input.m4a -ar 16000 -ac 1 output.wav
ffmpeg -i input.m4a -ar 16000 -ac 1 output.wav

```python

input_audio_path = '/Users/vlad/Library/CloudStorage/GoogleDrive-vladimir.smirnov@rhodaris.com/Shared drives/rhodaris working drive/Courses PMP maintenance 2026/Managing Conflict Upwards - Navigating Challenges with Superiors.m4a'
output_audio_name = "Managing Conflict Upwards_output16khz.wav"
output_transcription_name = "Managing Conflict Upwards transcription.txt"  # Specify the output file path


folder_path = os.path.dirname(input_audio_path)
output_audio_path = os.path.join(folder_path, output_audio_name)
output_file_path = os.path.join(folder_path, output_transcription_name)

meeting_language="en"

# Use subprocess to call ffmpeg for resampling the audio
# subprocess.run([
# 	"ffmpeg", "-i", input_audio_path, "-ar", "16000", "-ac", "1", output_audio_path
# ], check=True)
subprocess.run([
    "ffmpeg", "-y", "-vn", "-i", input_audio_path, "-ar", "16000", "-ac", "1", output_audio_path
], check=True)
```

# Specify Audio File Path
Define the path to the audio file that needs to be transcribed.

```python
# Specify Audio File Path
# audio_file_path = "/path/to/audio/file.wav"
# audio_file_path = "/Users/vlad/Dev/files/AUDIO-2025-04-18-17-40-23.m4a"
audio_file_path = output_audio_path
```

# Transcribe Audio
Use the Whisper model to transcribe the audio file and store the transcription.

```python
# Transcribe Audio
# result = model.transcribe(audio_file_path)
result = model.transcribe(audio_file_path, language=meeting_language)
transcription = result["text"]
```

```python
# Save Transcription to a Text File

with open(output_file_path, "w") as file:
    file.write(transcription)

print(f"Transcription saved to {output_file_path}")
```
